### Detección de duplicados y ventas negativas


No se encontraron registros duplicados en ninguno de los tres datasets.
En el dataset de ventas se identificaron 1.285 registros con valores negativos en `Weekly_Sales`, equivalentes aproximadamente al 0,305 % del total.
Estos registros podrían corresponder a devoluciones o ajustes contables. Se conservarán durante el análisis exploratorio hasta comprender mejor su origen y determinar si requieren algún tratamiento.


```python
negative_sales = sales[sales["Weekly_Sales"] < 0]

negative_sales.groupby("Store").size().sort_values(
    ascending=False
).head(10)
```

```python
negative_sales.groupby("Dept").size().sort_values(
    ascending=False
).head(10)
```

```python
negative_by_dept = (
    sales.assign(is_negative=sales["Weekly_Sales"] < 0)
    .groupby("Dept")["is_negative"]
    .agg(["sum", "count", "mean"])
    .sort_values("mean", ascending=False)
)

negative_by_dept.head(10)
```

### Análisis de ventas negativas por departamento

Se analizaron los registros con ventas negativas agrupándolos por departamento.

El departamento 47 presenta la mayor proporción de registros negativos, con aproximadamente un 39,32 % (254 de 646 observaciones).

También se identificaron proporciones elevadas en los departamentos 78 (14,04 %) y 77 (10,67 %).

Estos resultados sugieren que las ventas negativas no están distribuidas uniformemente entre los departamentos. Se requiere una investigación adicional para determinar su origen.

```python
sales_47 = sales[sales["Dept"] == 47]

sales_47["Weekly_Sales"].apply(
    lambda x: "Negative" if x < 0
    else "Positive" if x > 0
    else "Zero"
).value_counts()
```

### Distribución de ventas negativas en el departamento 47

Se investigó el departamento 47 debido a que presenta la mayor proporción de registros con ventas negativas (39,32 %).

El departamento aparece en 37 tiendas y contiene 646 observaciones:

- Ventas positivas: 373 (57,74 %).
- Ventas negativas: 254 (39,32 %).
- Ventas iguales a cero: 19 (2,94 %).

Las ventas negativas se encuentran distribuidas entre múltiples tiendas. Además, la media del departamento es negativa (-7,68), mientras que su mediana es positiva (17,985).

Estos resultados indican un comportamiento particular que requiere consideración durante el análisis y la preparación de los datos. No se eliminarán registros sin determinar previamente su origen.
