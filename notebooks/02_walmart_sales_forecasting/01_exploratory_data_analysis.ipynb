# Ahora vamos a analizar las ventas por departamento

```python
sales_by_dept = (
    sales.groupby("Dept")["Weekly_Sales"]
    .agg(["mean", "median", "sum", "count"])
    .sort_values("mean", ascending=False)
)

sales_by_dept.head(10)
```

# Siguiente paso: visualizar los departamentos

```python
plt.figure(figsize=(12, 5))

sns.barplot(
    data=sales_by_dept.head(10).reset_index(),
    x="Dept",
    y="mean"
)

plt.title("Top 10 Departments by Average Weekly Sales")
plt.xlabel("Department")
plt.ylabel("Average Weekly Sales")

plt.show()
```

```python
top_departments = sales_by_dept.sort_values(
    "sum", ascending=False
)

top_departments.head(10)
```

# Para una mejor comprensión, vemos las ventas por tienda

```python
sales_by_store = (
    sales.groupby("Store")["Weekly_Sales"]
    .agg(["mean", "median", "sum", "count"])
    .sort_values("sum", ascending=False)
)

sales_by_store.head(10)
```

```python
sales_by_store = sales_by_store.merge(
    stores,
    on="Store",
    how="left"
)

sales_by_store.head(10)
```

```python
sales_by_store = (
    sales.groupby("Store")["Weekly_Sales"]
    .agg(["mean", "median", "sum", "count"])
    .sort_values("sum", ascending=False)
)

sales_by_store.head(10)
```

# Investiguemos la relación entre tamaño y ventas

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=sales_by_store,
    x="Size",
    y="mean",
    hue="Type",
    s=100
)

plt.title("Store Size vs Average Weekly Sales")
plt.xlabel("Store Size")
plt.ylabel("Average Weekly Sales")

plt.show()
```

# Vamos a hacer el coeficiente de correlación de Pearson para establecer mejor la relación entre tamaño de tienda y ventas semanales

```python
correlation_matrix = sales_by_store[
    ["mean", "median", "sum", "count", "Size"]
].corr()

correlation_matrix
```

# Ahora lo visualizamos con el mapa de calor

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Store Sales Correlation Matrix")

plt.show()
```

# Calculamos las correlaciones entre tamaño y ventas separadas por tipo de tienda

```python
correlation_by_type = (
    sales_by_store.groupby("Type")
    .apply(lambda group: group[["Size", "mean"]].corr().iloc[0, 1])
)

correlation_by_type
```

## Al analizar la correlación entre el tamaño de las tiendas y sus ventas promedio según el tipo de establecimiento, se observan diferencias entre categorías. Las tiendas tipo A presentan una correlación positiva de 0.770, las tipo B de 0.561 y las tipo C de 0.112. Esto sugiere que la relación entre superficie y ventas no es uniforme entre los distintos tipos de tiendas.
